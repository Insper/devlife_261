## Combinando os condicionais

`and` `or` `not`

```python
num_patas = 2
num_asas = 2
nota_ronronar = 9
```

```python
if num_patas == 0:
    print("Peixe")
if num_patas == 2 or num_patas == 4:
    if num_asas == 2: 
        print("Ave")
    elif num_asas == 0: 
        if nota_ronronar >= 3:
            print("Gato")
        else: 
            print("Cachorro")
```
