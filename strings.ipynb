## Split

```python
blocos = frase.split()
```

```python
blocos
```

```python
for palavra in frase.split():
    print(palavra)
```

```python
banana = "a banana de pijama"

saida = []
for b in banana.split():
    if b!="banana":
        saida.append(b)
    else:
        saida.append(b.replace("a", "o"))
    
saida
```

## Função join

```python
" ".join(saida)
```

```python
saida
```

## Usando o operador +

```python
saida2 = ""
for p in saida: 
    saida2 += p 
    saida2 += " "
print(saida2)
```
