# Relembrando o for

```python
lista_n = range(0, 10)
```

```python
for i in range(10, -1, -2):
    print(i)
```

```python
nomes = [
    "João",
    "Maria",
    "Francisco",
    "Ana",
    "Pedro",
    "Beatriz",
    "Miguel",
    "Alice",
    "Gabriel",
    "Helena"
]
```

```python
for e in nomes:
    print(e)
```

```python
for i in  range(len(nomes)):
    print(f"{i}:{nomes[i]}")
```

```python
enumerate(nomes)
```

```python
for i, nome in enumerate(nomes):
    print(f"{i+1}:{nome}")
```

```python
list(enumerate(nomes))
```
